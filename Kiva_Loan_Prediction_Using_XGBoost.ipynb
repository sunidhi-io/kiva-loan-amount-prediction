```python
# Select useful features :-
df_model = df[['loan_amount',
             'funded_amount',
             'term_in_months',
             'sector',
             'country',
             'lender_count']]
```

```python
# View selected dataset :-
df_model.head()
```

```python
# ~~~~~~~~~~~~~~~~ Feature Engineering ~~~~~~~~~~~~~~~~

# Encode categorical columns :-
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

df_model['sector'] = le.fit_transform(df_model['sector'])
df_model['country'] = le.fit_transform(df_model['country'])
```

```python
# Check datatypes :-
df_model.info()
```

```python
# Save Clean Dataset :-
df.to_csv("clean_data.csv",index=False)
```

```python
# Download Clean Dataset :-
from google.colab import files
files.download("clean_data.csv")
```

```python
# Regression :-

# Split features and target :-
x = df_model.drop('loan_amount',axis=1)
y = df_model['loan_amount']
```

```python
# Split data :-
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# Check Train Size :-
x_train.shape,x_test.shape
```

```python
# Import XGBoost :-
from xgboost import XGBRegressor

# Create XGBoost model :-
model = XGBRegressor(
    n_estimators = 100,
    random_state = 42
)
```

```python
# Train model :-
model.fit(x_train,y_train)
```

```python
# Make prediction :-
y_pred = model.predict(x_test)
```

```python
# Import metrics :-
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Model evaluation :-
print('Mean Absolute Error:',mean_absolute_error(y_test, y_pred))
print('Mean Squared Error:',mean_squared_error(y_test, y_pred)**0.5)
print('R2 Score:',r2_score(y_test, y_pred))

```

```python
# Plot actual vs predicted :-
plt.figure(figsize=(8,5))
plt.scatter(y_test, y_pred)
plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], linestyle='--', color='red')
plt.title('Actual vs Predicted Loan Amount')
plt.xlabel('Actual Loan Amount')
plt.ylabel('Predicted Loan Amount')
```

```python
# Compare results :-
results = pd.DataFrame({ 'Actual': y_test, 'Predicted': y_pred })
print(results)
```
